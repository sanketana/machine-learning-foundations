## Exercise 1 · Three scenarios — solution

**A (cricket):** (i) label = will the batter score 50+ (yes/no); (ii) features = balls faced, strike rate, opposition, venue — past-innings stats; (iii) **supervised** — history provides the answers.

**B (keyboard):** (i) label = the next word; (ii) features = the words typed so far (context); (iii) **supervised** — every typed sentence supplies "what word actually came next" examples.

**C (listener segments):** (i) nothing is predicted — the job is to *find groups*; (ii) features = play history (artists, genres, times, skips); (iii) **unsupervised** — no answer column exists.

📋 *Acceptable alternatives:* In B, "predict which word the user will tap" is fine. Students often call C supervised because "the app knows a lot" — the corrective question is "knows a lot of *answers to what*? Show me the answer column." If they can't, it's unsupervised.

## Exercise 3 · Be the rules — example solution

```python
def will_i_reply_quickly(sender_is_friend, message_has_question, hour_of_day):
    """Example solution — any 2-3 sensible rules are correct."""
    if 0 <= hour_of_day < 7:
        return False                 # asleep: nobody gets a fast reply
    if sender_is_friend and message_has_question:
        return True                  # friends' questions feel urgent
    if message_has_question:
        return True                  # questions nag at you regardless
    return False


print("Friend asks a question at 5pm: ", will_i_reply_quickly(True,  True,  17))
print("Stranger, no question, 11pm:   ", will_i_reply_quickly(False, False, 23))
print("Friend, no question, 7am:      ", will_i_reply_quickly(True,  False, 7))
```

📋 *Grading notes:* any rules count as long as the function runs on all three test cases and returns a boolean. The valuable part is the failure description — e.g. "my rules say I ignore strangers, but I'd reply fast to a stranger returning my lost bag." Look for the *shape* of the insight: hand-written rules can't anticipate context. That is the whole argument for ML, in the student's own words.

## Exercise 4 · The memorizer — solution and what to look for

Adding any unseen title produces `??? (never seen this title)` — the model has no opinion about anything outside its dictionary.

**Reference reflection:** "The memorizer is just storing my past answers, so it can only repeat them — it never learned *why* I liked those movies. Memorizing my ratings looks perfect on old movies but tells it nothing about new ones. A model that learned my taste would use properties of the movies — genre, pace, length, actors — as features, so the pattern it found would carry over to titles it has never seen."

📋 *Look for:* (1) memorizing = storing answers, learning = finding a pattern; (2) the failure is *on new data specifically*; (3) the fix is features that generalize, not more titles in the dictionary. If the reflection says "it needs more data" — that's the misconception to correct in homework review: a bigger answer key is still an answer key.
