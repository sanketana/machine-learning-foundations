## Part 1 · The machine that follows rules

Before machine learning, there was only one way to make a computer decide anything: **a human writes the rules.**

Let's be that human. Below is a function that decides whether a song will be a hit, using rules we made up ourselves.

````python
def is_hit_song(danceability, artist_followers_millions, length_minutes):
    """Decide whether a song will be a hit, using hand-written rules.

    danceability is a score from 0 (a lecture) to 1 (impossible not to move).
    """
    if danceability > 0.8 and length_minutes < 4:
        return True            # short and catchy usually works
    if artist_followers_millions > 50:
        return True            # superstars get streamed no matter what
    return False               # everything else: probably not


# Three songs to test our rules on
print("Catchy newcomer:  ", is_hit_song(danceability=0.9, artist_followers_millions=2,  length_minutes=3.2))
print("Superstar ballad: ", is_hit_song(danceability=0.4, artist_followers_millions=80, length_minutes=5.0))
print("Quiet indie track:", is_hit_song(danceability=0.5, artist_followers_millions=1,  length_minutes=4.5))
````

## Part 2 · The flip

Here is how Netflix knows what to recommend to you, even though no programmer at Netflix has ever heard of you:

```
Traditional programming:   rules + data     →  answers
Machine learning:          data + answers   →  rules
```

In machine learning we hand the computer **examples with answers** — millions of "this person watched X and then chose Y" records — and the computer works out the rules itself.

The rules it finds are called a **model**. That word will appear in every session from now on.

### The words we'll use every session

## Part 3 · Memorizing is not learning

Imagine a student who prepares for a test by memorizing the answer key from last year — every question, every answer, perfectly.

On last year's questions they score 100%. Are they ready for *this* year's test?

Let's build that student as code. Here is a "model" that has memorized which songs were hits:

````python
# A "model" that memorizes the answer key: song title -> was it a hit?
answer_key = {
    "Midnight Drive": "yes",
    "Gravity":        "yes",
    "Paper Boats":    "no",
    "Neon Heart":     "yes",
    "Last Bus Home":  "no",
    "Static":         "no",
}

def memorizing_model(title):
    if title in answer_key:
        return answer_key[title]
    return "??? (never seen this song)"


# Quiz it on every song it has seen:
correct = 0
for title, truth in answer_key.items():
    prediction = memorizing_model(title)
    if prediction == truth:
        correct += 1
    print(f"{title:15s} predicted: {prediction:4s}  actual: {truth}")

print(f"\nScore on songs it has seen: {correct}/{len(answer_key)} — perfect!")
````

````python
# Now four songs it has NEVER seen — including some certain hits:
new_songs = ["Golden Hour", "Counting Stars", "Monsoon", "Glass City"]

for title in new_songs:
    print(f"{title:15s} predicted: {memorizing_model(title)}")
````

### ✏️ The phrase to keep

The memorizing model scored **100% on songs it had seen** and was **useless on every song it hadn't**. It never found any pattern connecting danceability or followers to success — it just stored answers.

## Part 4 · Two kinds of learning

| | The dataset has… | The model's job | Example |
|---|---|---|---|
| **Supervised** | an answer column (a label) | predict the label for new rows | predict `became_hit` from song features |
| **Unsupervised** | no answer column | find natural groups or structure | group listeners by taste, with no categories given |

Quick test for any example: *is there an answer column?* Yes → supervised. No → unsupervised.
