## Step 1 · Meet the data

Load `housing.csv` and scatter house size against price.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

homes = pd.read_csv("../../../datasets/secondary/housing.csv")

feature = "area_sqft"
label = "price_lakhs"

print("homes:", homes.shape[0])
homes[[feature, label]].head()
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(homes[feature], homes[label], alpha=0.4)
plt.xlabel("area (square feet)")
plt.ylabel("price (lakhs)")
plt.title("Each dot is one house")
plt.show()
```

## Step 2 · Draw a line by eye

Pick a **slope** (lakhs of price per extra square foot — this is a small number, since one square foot barely moves the price) and an **intercept**. Run the cell, see your line, and adjust until it looks like a good fit.

```python
# ✏️ Pick a slope and intercept by eye. (A reasonable teacher guess below.)
my_slope = 0.07       # lakhs of price per extra square foot
my_intercept = 15.0   # price at zero area (just where the line starts)

x_line = np.array([homes[feature].min(), homes[feature].max()])
plt.figure(figsize=(7, 5))
plt.scatter(homes[feature], homes[label], alpha=0.4, label="houses")
plt.plot(x_line, my_slope * x_line + my_intercept, color="red", linewidth=2, label="my line")
plt.xlabel("area (square feet)")
plt.ylabel("price (lakhs)")
plt.legend()
plt.show()
```

## Step 3 · Let sklearn draw it

```python
from sklearn.linear_model import LinearRegression

X = homes[["area_sqft"]]   # double brackets → one-column table
y = homes["price_lakhs"]

# Fit LinearRegression on the single feature area_sqft.
model = LinearRegression()
model.fit(X, y)

sk_slope = model.coef_[0]
sk_intercept = model.intercept_
print(f"sklearn slope:     {sk_slope:.4f}  (lakhs per square foot)")
print(f"sklearn intercept: {sk_intercept:.2f}  (price at zero area)")
```

```python
# Overlay your line (red) and sklearn's (green).
plt.figure(figsize=(7, 5))
plt.scatter(homes[feature], homes[label], alpha=0.3, label="houses")
plt.plot(x_line, my_slope * x_line + my_intercept, color="red", linewidth=2, label="my line")
plt.plot(x_line, sk_slope * x_line + sk_intercept, color="green", linewidth=2, label="sklearn line")
plt.xlabel("area (square feet)")
plt.ylabel("price (lakhs)")
plt.legend()
plt.show()
```

### ✏️ Read the line

1. “In this data, each extra **100** square feet is associated with about ____ lakhs more in price.” *(Hint: multiply the slope by 100.)*
2. Is the fit **tight** (dots hug the line) or **loose** (dots scatter widely off it)? What might that tell you about whether *size alone* decides a house’s price?

*Your answers:*

1. In this data, each extra 100 sq ft is associated with about **6.5 lakhs** more (slope ≈ 0.066 × 100). *(teacher: exact value depends on nothing random here — there is no split — so it should be ~0.0655.)*
2. Loose — the dots scatter well off the line. Size clearly matters (upward trend) but plainly isn’t the whole story; bedrooms, age, and location must matter too. This looseness is exactly why we’ll want (a) a *number* for how loose — Sessions 6 and 8 — and (b) *more features* — Session 7.

## Step 4 · Predict and interpret

Use the line to price a house of a size you choose.

```python
area = 1500   # ✏️ pick any house size in square feet

by_model = model.predict(pd.DataFrame({"area_sqft": [area]}))[0]
by_hand = sk_slope * area + sk_intercept
print(f"predicted price for {area} sqft: {by_model:.1f} lakhs")
print(f"by hand (slope*area+intercept): {by_hand:.1f} lakhs")
```
