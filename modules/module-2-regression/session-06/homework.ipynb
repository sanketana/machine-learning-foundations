## Step 1 · Cost of two lines

Compute the cost (average squared miss) of your by-eye line and of sklearn's, and see which is lower.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

homes = pd.read_csv("../../../datasets/secondary/housing.csv")
feature, label = "area_sqft", "price_lakhs"
x = homes[feature].values
y = homes[label].values

model = LinearRegression().fit(homes[[feature]], y)
sk_slope, sk_intercept = model.coef_[0], model.intercept_

def cost(slope, intercept):
    predicted = slope * x + intercept
    return np.mean((y - predicted)**2)

# ✏️ Your by-eye housing line from Session 5 (or keep this guess).
my_slope, my_intercept = 0.05, 20.0

# ✏️ TODO: nothing to write here — read the two costs printed below.
print(f"cost of MY line:      {cost(my_slope, my_intercept):.1f}")
print(f"cost of SKLEARN line: {cost(sk_slope, sk_intercept):.1f}")
```

## Step 2 · Draw the bowl

Sweep the slope across a range (intercept fixed at sklearn's) and plot each slope's cost. House-price slopes are small numbers, so the range is small.

```python
slopes = np.linspace(0, 0.14, 71)
costs = [cost(s, sk_intercept) for s in slopes]

plt.figure(figsize=(7, 5))
plt.plot(slopes, costs, color="purple")
plt.xlabel("slope (lakhs per square foot)")
plt.ylabel("cost (average squared miss)")
plt.title("The bowl for house prices")
plt.show()
```

## Step 3 · Find the floor and check `.fit()`

```python
best_i = int(np.argmin(costs))
best_slope = slopes[best_i]

print(f"bottom of the bowl at slope ≈ {best_slope:.3f}")
print(f"sklearn .fit() slope       = {sk_slope:.3f}")

plt.figure(figsize=(7, 5))
plt.plot(slopes, costs, color="purple")
plt.scatter([best_slope], [costs[best_i]], color="red", zorder=5, label="bottom (best line)")
plt.axvline(sk_slope, color="green", linestyle="--", label="sklearn .fit() slope")
plt.xlabel("slope"); plt.ylabel("cost")
plt.legend(); plt.show()
```
