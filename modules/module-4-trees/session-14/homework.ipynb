## Part 2 · Fit a depth-1 tree on the full data

Fit `max_depth=1` on the five habit features. Report the root split and test accuracy.

```python
X = full[habits]
y = full['passed']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)

# ✏️ create and fit a DecisionTreeClassifier with max_depth=1, random_state=0
stump = DecisionTreeClassifier(max_depth=1, random_state=0)  # (leave as-is or rebuild yourself)
stump.fit(X_train, y_train)
print(export_text(stump, feature_names=habits))
print('test accuracy:', round(stump.score(X_test, y_test), 3))
```
