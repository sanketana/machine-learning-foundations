## Step 1 · The baseline that lies

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score

fraud = pd.read_csv("../../../datasets/secondary/fraud_transactions.csv")
X = fraud.drop(columns=["is_fraud", "transaction_id"]).values
y = fraud["is_fraud"].values
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)
scaler = StandardScaler().fit(X_train)
X_train_s = scaler.transform(X_train)
X_test_s = scaler.transform(X_test)

# ✏️ TODO: build the 'always legit' baseline; report its accuracy AND recall.
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
bp = baseline.predict(X_test)
print("baseline accuracy:", round(accuracy_score(y_test, bp), 3))
print("baseline recall:  ", round(recall_score(y_test, bp, zero_division=0), 3), " (fraction of real fraud caught)")
```

## Step 2 · A real model, judged properly

```python
# ✏️ TODO: fit logistic regression; report accuracy, precision, recall.
model = LogisticRegression(max_iter=1000).fit(X_train_s, y_train)
pred = model.predict(X_test_s)
print("model accuracy :", round(accuracy_score(y_test, pred), 3))
print("model precision:", round(precision_score(y_test, pred), 2))
print("model recall   :", round(recall_score(y_test, pred), 2))
```
