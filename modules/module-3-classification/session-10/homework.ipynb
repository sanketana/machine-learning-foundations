The point isn't just "which wins" — it's that logistic regression hands back a **probability** per student, something KNN's hard vote never did.

## Step 1 · Load, scale, split

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

students = pd.read_csv("../../../datasets/anchor/student_habits.csv")
habits = ["study_hours_per_week", "attendance_pct", "sleep_hours_per_night",
          "screen_time_hours_per_day", "practice_sessions_per_week"]
X = students[habits].values
y = students["passed"].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)
scaler = StandardScaler().fit(X_train)
X_train_s = scaler.transform(X_train)
X_test_s = scaler.transform(X_test)
```

## Step 2 · Fit both models on the same split

Report **test accuracy** for each — a named metric, and a fair comparison because they share the exact same train/test rows.

```python
# ✏️ TODO: fit logistic regression and report its test accuracy.
logit = LogisticRegression(max_iter=1000).fit(X_train_s, y_train)
print("logistic regression test accuracy:", round(logit.score(X_test_s, y_test), 3))

# ✏️ TODO: fit KNN (k=15) on the same split and report its test accuracy.
knn = KNeighborsClassifier(n_neighbors=15).fit(X_train_s, y_train)
print("KNN (k=15) test accuracy:        ", round(knn.score(X_test_s, y_test), 3))
```

## Step 3 · Read the probabilities

Logistic regression's edge is the **probability**. Print P(pass) for three test students: one the model is confident *passes*, one it's confident *fails*, and one it's genuinely **unsure** about (closest to 0.5).

```python
p_pass = logit.predict_proba(X_test_s)[:, 1]

i_conf_pass = int(np.argmax(p_pass))               # highest P(pass)
i_conf_fail = int(np.argmin(p_pass))               # lowest P(pass)
i_unsure    = int(np.argmin(np.abs(p_pass - 0.5))) # closest to 0.5

print("confident PASS  P(pass) =", round(float(p_pass[i_conf_pass]), 3))
print("confident FAIL  P(pass) =", round(float(p_pass[i_conf_fail]), 3))
print("least sure      P(pass) =", round(float(p_pass[i_unsure]), 3), " <-- closest to 0.5")
```
